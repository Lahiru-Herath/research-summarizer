```python
from src.router import DomainRouter

router = DomainRouter()
print("Router ready.")
```

```python
from src.output_generator import StructuredOutputGenerator

generator = StructuredOutputGenerator(save_outputs=True)
print("Generator ready.")
```

```python
general_paper = {
    "title": "Attention Is All You Need",
    "text": (
        "Attention Is All You Need [SEP] "
        "The dominant sequence transduction models are based on complex recurrent "
        "or convolutional neural networks that include an encoder and a decoder. "
        "The best performing models also connect the encoder and decoder through "
        "an attention mechanism. We propose a new simple network architecture, "
        "the Transformer, based solely on attention mechanisms, dispensing with "
        "recurrence and convolutions entirely. Experiments on two machine "
        "translation tasks show these models to be superior in quality while "
        "being more parallelizable and requiring significantly less time to train."
    )
}

result_general = generator.generate(
    text   = general_paper["text"],
    title  = general_paper["title"],
    router = router,
)
```

```python
print("=" * 60)
print(f"Title      : {result_general['title']}")
print(f"Domain     : {result_general['domain']}")
print(f"Confidence : {result_general['confidence']}")
print(f"Model used : {result_general['model_used']}")
print(f"\nSummary:\n{result_general['summary']}")
print(f"\nKey Findings:")
for f in result_general["key_findings"]:
    print(f"  • {f}")
print(f"\nMethodology:\n  {result_general['methodology']}")
print(f"\nLimitations:")
for l in result_general["limitations"]:
    print(f"  • {l}")
print(f"\nFuture Work:")
for fw in result_general["future_work"]:
    print(f"  • {fw}")
print("=" * 60)
```
